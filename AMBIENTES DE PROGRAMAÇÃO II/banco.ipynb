```python
import sqlite3
import os
```

```python
if os.path.exists('unoeste.db'):
    os.remove('unoeste.db')
    print('Bnaco Removido')
else:
    print('Ambiente Pronto')
```

```python
con = sqlite3.connect('unoeste.db')
```

```python
cur = con.cursor()
```

```python
sql_create = """
CREATE TABLE alunos
(
    RA inter not null primary key,
    Nome varchar(100)

)
"""

cur.execute(sql_create)
```

```python
sql_insert = """
INSERT INTO alunos
values (262619890,'Pedro')
"""

cur.execute(sql_insert)
```

```python
con.commit()
```

```python
sql_select = """
SELECT * FROM alunos
"""
cur.execute(sql_select)
dados = cur.fetchall()
print(dados)
```
